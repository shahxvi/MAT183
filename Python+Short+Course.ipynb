## While Loop

Use a while loop to generate and print a table of conversion of °F to °C for F = {0, 5, ..., 40}.

**Hint:** $C = 5(F - 32)/9$

```python
print("F\tC")
F = 0
while F <= 40:
  print(F, "\t", 5 * (F - 32) / 9)
  F += 5
```

## List

Create a list called F contains temperature values in °F ranges from 0 to 40 with step size 5 and print it.

## For Loop

Rewrite the while loop program to use for loop.

```python
print("F\tC")
for f in F:
  print(f,"\t", 5*(f - 32)/9)
```

## List 2

Rewrite the for loop program to store the conversion values into a list called C.

Print C.

```python
C = []
for f in F:
  C.append(5*(f - 31)/9)
print(F)
print(C)
```

## List Comprehension

Rewrite the program to use list comprehension to populate C.

Print C.

```python
C = [(5 * (temp - 32) / 9) for temp in F]
print(C)
```

## Lambda

Rewite the program to use lambda.

```python
calcC = lambda x: 5*(x - 32)/9
for f in F:
  print(calcC(f))
```

## Map

Rewrite the program to use map.

```python
C =  map(calcC, F)
print(list(C))
```

## Nested List

Create a list called table of (F, C) using zip.

```python
fc = list(zip(F, C))
print(fc)
```

## Functions

Single argument

```python
def to_C(F):
    '''Return the conversion of F to C'''
    return 5*(F - 32)/9
```

```python
to_C(0)
```

Multiple arguments

```python
def multiply(a, b):
    '''Return the axb'''
    return a*b
```

```python
multiply(2,3)
```

Multiple return values

```python
def add_multiply(a, b):
    '''Return a+b, a*b'''
    return a+b, a*b
```

```python
add_multiply(2,3)
```
